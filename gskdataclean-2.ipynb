```python
import pandas as pd

s = df_clean["Admission_Date"].astype("string").str.strip()
dates = pd.Series(pd.NaT, index=s.index, dtype="datetime64[ns]")
matched = s.isna()

formats = [
    (r"\d{4}", "%Y"),
    (r"\d{4}/\d{2}/\d{2}", "%Y/%m/%d"),
    (r"\d{4}-\d{2}-\d{2}", "%Y-%m-%d"),
    (r"\d{2}/\d{2}/\d{4}", "%d/%m/%Y"),
    (r"\d{2}\.\d{2}\.\d{4}", "%d.%m.%Y"),
    (r"[A-Za-z]+ \d{1,2}, \d{4}", "%B %d, %Y"),
    (r"\d{2}-[A-Za-z]{3}-\d{4}", "%d-%b-%Y")
]

for pattern, fmt in formats:
    mask = s.str.fullmatch(pattern, na=False)
    parsed = pd.to_datetime(s[mask], format=fmt, errors="coerce")
    dates.loc[mask] = parsed
    matched |= mask

# Handle numeric hyphen dates when the day/month order is clear
mask = s.str.fullmatch(r"\d{2}-\d{2}-\d{4}", na=False)
parts = s[mask].str.split("-", expand=True)
first = parts[0].astype(int)
second = parts[1].astype(int)

day_first = mask.copy()
day_first.loc[mask] = (first > 12) | (second <= 12)

month_first = mask.copy()
month_first.loc[mask] = (second > 12) & (first <= 12)

# Parse only unambiguous dates
dates.loc[day_first[day_first].index] = pd.to_datetime(
    s[day_first], format="%d-%m-%Y", errors="coerce"
)
dates.loc[month_first[month_first].index] = pd.to_datetime(
    s[month_first], format="%m-%d-%Y", errors="coerce"
)

matched |= day_first | month_first

# Report results; do not replace the original column yet
remaining = s.notna() & dates.isna()

print("Total rows:", len(s))
print("Original missing values:", s.isna().sum())
print("Dates converted:", dates.notna().sum())
print("Remaining values:", remaining.sum())
print("\nRemaining examples:")
print(s[remaining].head(20))
```
